```python
# Basic setup
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

df = pd.read_csv('/content/inr_4_newold_dataset.csv')

#Dataset overview
print(" Dataset Loaded \n")

print("Shape of dataset:", df.shape)

print("\n Column Names:")
print(df.columns.tolist())

print("\nℹ Data Info:")
df.info()

print("\n First 5 Rows:")
display(df.head())

print("\n Last 5 Rows:")
display(df.tail())

# Descriptive statistics
print("\n Numerical Summary Statistics:")
display(df.describe())
print("\nCategorical Summary Statistics:")
display(df.describe(include=['object', 'category']))
```

Data Cleaning


```python

print("\nTarget Classes:", df['Class'].unique())
print("\nClass Distribution:")
print(df['Class'].value_counts())

print("\nFirst 5 Rows:")
display(df.head())
```

```python
# Fix Data Types

df['Class'] = df['Class'].astype('category')

print(" Data Types Summary:")
print(df.dtypes.value_counts())

print("\n DataFrame Info:")
df.info()

print("\n Categorical Columns:", df.select_dtypes(include='category').columns.tolist())
print(" Numeric Columns:", df.select_dtypes(include=['float64', 'int64']).columns.tolist()[:10], "...")  # preview
```

```python
# Handle Missing Values

missing = df.isnull().sum()

print("Missing Values Summary:\n")
if missing.sum() == 0:
    print(" No missing values found in any column.")
else:
    display(missing[missing > 0])

possible_missing_indicators = ['?', 'NA', 'N/A', 'na', 'null', 'None', '']
for col in df.select_dtypes(include='object').columns:
    mask = df[col].isin(possible_missing_indicators)
    if mask.any():
        print(f"Hidden missing indicators found in column: {col}")
        display(df.loc[mask, col])

print("\n Missing value check complete.")
```

```python
#Remove Duplicates

duplicates = df.duplicated().sum()
print(f"Duplicate Rows Found: {duplicates}")

if duplicates > 0:
    df = df.drop_duplicates()
    print(f"Duplicates removed. New shape: {df.shape}")
else:
    print("No duplicate rows found. Data is unique.")
```

```python
#Detect Outliers

num_cols = df.select_dtypes(include=['float64', 'int64']).columns

print(" Performing outliers...\n")

for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    before = df[col].copy()
    df[col] = df[col].clip(lower, upper)

    changed = (before != df[col]).sum()
    print(f"{col}: {changed} values")

print("\n Outlier completed")

```

```python
# Fix Inconsistent Values
num_cols = df.select_dtypes(include=['float64', 'int64']).columns
print("Checking data quality...\n")
cols_with_neg = []

for col in num_cols:
    min_val = df[col].min()
    if min_val < 0:
        print(f"{col}: Negative values found (min={min_val})")
        cols_with_neg.append(col)
    else:
        print(f"{col}: ok")

if cols_with_neg:
    print("\n Fixing negative values...\n")
    for col in cols_with_neg:
        neg_count = (df[col] < 0).sum()
        df.loc[df[col] < 0, col] = 0
        print(f"{col}: {neg_count} negative values replaced with 0")
    print("\n Negative value correction complete ")
else:
    print("\n No negative values found")

print("\n Data quality check & fix complete.")

```

```python
#EDA
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/inr_4_newold_dataset.csv')
print("Dataset Loaded Successfully!")
print("Shape:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes.value_counts())

print("\nDataset Info:")
df.info()

print("\nDescriptive Statistics (Numerical):")
display(df.describe().T.round(3))

print("\nDescriptive Statistics (Categorical):")
display(df.describe(include=['object', 'category']))

print("\nMissing Values Summary:")
print(df.isnull().sum())

print("\nDuplicate Rows Found:", df.duplicated().sum())
if df.duplicated().sum() == 0:
    print("No duplicate rows found. Data is unique.")

print("\nPerforming Outlier Check...")
for col in df.select_dtypes(include=['float64', 'int64']).columns:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lower, upper = q1 - 1.5*iqr, q3 + 1.5*iqr
    outliers = ((df[col] < lower) | (df[col] > upper)).sum()
    print(f"{col}: {outliers} outliers")

print("\nOutlier check complete.")

print("\nChecking Data Quality...")
for col in df.select_dtypes(include=['float64', 'int64']).columns:
    if (df[col] < 0).any():
        print(f"{col}: Negative values found (min={df[col].min()})")
    else:
        print(f"{col}: OK")

fix_cols = ['X1', 'CN1', 'CN2']
for col in fix_cols:
    neg_count = (df[col] < 0).sum()
    if neg_count > 0:
        df.loc[df[col] < 0, col] = 0
        print(f"{col}: {neg_count} negative values replaced with 0")

print("\nNegative value correction complete.")
print("\nData Quality Check & Fix Complete.")

if 'Class' in df.columns:
    plt.figure(figsize=(6,4))
    df['Class'].value_counts().plot(kind='bar', color='skyblue', edgecolor='black')
    plt.title('Class Distribution')
    plt.xlabel('Class')
    plt.ylabel('Count')
    plt.show()
else:
    print("\n'Class' column not found for visualization.")

plt.figure(figsize=(10,8))
sns.heatmap(df.corr(numeric_only=True), cmap='coolwarm', center=0)
plt.title("Feature Correlation Heatmap")
plt.show()

if 'Class' in df.columns:
    sns.pairplot(df, vars=['R1', 'G1', 'B1', 'W1'], hue='Class', corner=True)
else:
    sns.pairplot(df, vars=['R1', 'G1', 'B1', 'W1'], corner=True)
plt.show()

print("\nEDA complete")
```
